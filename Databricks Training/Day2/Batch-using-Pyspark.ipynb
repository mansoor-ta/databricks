```python
input_path="/Volumes/dev/naval/raw/financial_dataset.csv"
catalog="dev"
schema="mansoor"
table_name="financial_pyspark"
```

```python
def add_columns(df):
    return df.withColumn("ingestion_date",current_timestamp()).withColumn("path",col("_metadata.file_name"))
```

```python
df=(spark.read.csv(input_path,header=True,inferSchema=True))
df1=add_columns(df)
df1.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.{table_name}")
```

```python
df1.display()
df1.filter("amountusd > 3000").display()
df1.orderBy(col("AmountUSD").desc()).display()
df1.groupBy("branch").count().display()
df1.groupBy("branch").agg(sum("amountusd")).display()
```
